## 5.1 User Health Profile and BMI Assessment

ในส่วนนี้เพิ่มการประเมินข้อมูลพื้นฐานของผู้ใช้ เช่น น้ำหนัก ส่วนสูง เพศ และเส้นรอบเอว
เพื่อคำนวณ BMI และประเมินความเสี่ยงเบื้องต้น ซึ่งช่วยให้ระบบสามารถให้คำแนะนำแบบ personalized ได้มากขึ้น

```python
# Example user profile
# สามารถเปลี่ยนค่าได้ตามผู้ใช้จริง
weight_kg = 70
height_cm = 170
gender = 'male'   # 'male' or 'female'
waist_cm = 88

height_m = height_cm / 100
bmi = weight_kg / (height_m ** 2)

def bmi_category_asian(bmi):
    if bmi < 18.5:
        return 'Underweight'
    elif bmi < 23.0:
        return 'Normal'
    elif bmi < 25.0:
        return 'Overweight'
    elif bmi < 30.0:
        return 'Obese Level 1'
    else:
        return 'Obese Level 2'

def waist_risk(gender, waist_cm):
    gender = gender.lower()
    if gender == 'male':
        return 'High Risk' if waist_cm > 90 else 'Normal'
    elif gender == 'female':
        return 'High Risk' if waist_cm > 80 else 'Normal'
    return 'Unknown'

bmi_group = bmi_category_asian(bmi)
waist_status = waist_risk(gender, waist_cm)

protein_min = 1.2 * weight_kg
protein_max = 1.6 * weight_kg

user_profile_df = pd.DataFrame([
    {
        'weight_kg': weight_kg,
        'height_cm': height_cm,
        'gender': gender,
        'waist_cm': waist_cm,
        'bmi': round(bmi, 2),
        'bmi_category': bmi_group,
        'waist_risk': waist_status,
        'protein_min_g_day': round(protein_min, 2),
        'protein_max_g_day': round(protein_max, 2)
    }
])

user_profile_df
```

จากข้อมูลผู้ใช้ตัวอย่าง ระบบสามารถคำนวณ BMI และช่วงโปรตีนที่แนะนำต่อวันได้
โดยอ้างอิงสูตร BMI และคำแนะนำโปรตีนจากเอกสารมาตรฐาน ซึ่งช่วยให้การประเมินอาหารในขั้นต่อไปสอดคล้องกับข้อมูลของผู้ใช้มากขึ้น

```python
print('BMI =', round(bmi, 2))
print('BMI Category =', bmi_group)
print('Waist Risk =', waist_status)
print('Recommended Protein Range =', round(protein_min, 2), '-', round(protein_max, 2), 'g/day')
```

```python
df_model = df_eda.copy()
before_rows = df_model.shape[0]

df_model = df_model.drop_duplicates()
after_rows = df_model.shape[0]

print('Rows before:', before_rows)
print('Rows after removing duplicates:', after_rows)
```

```python
for col in nutrition_cols:
    df_model[col] = df_model[col].fillna(df_model[col].median())

df_model.isnull().sum()
```

```python
standard_df = pd.read_csv("../data/knowledge/standard_df.csv")
dga_standard_df = pd.read_csv("../data/knowledge/dga_standard_df.csv")

THAI_RDI_SODIUM = standard_df.loc[
    standard_df["metric"] == "Sodium", "recommended_value"
].iloc[0]

THAI_RDI_CHOLESTEROL = standard_df.loc[
    standard_df["metric"] == "Cholesterol", "recommended_value"
].iloc[0]

THAI_RDI_SAT_FAT = standard_df.loc[
    standard_df["metric"] == "Saturated Fat", "recommended_value"
].iloc[0]

THAI_RDI_FAT = standard_df.loc[
    standard_df["metric"] == "Total Fat", "recommended_value"
].iloc[0]

THAI_RDI_CARBS = standard_df.loc[
    standard_df["metric"] == "Total Carbohydrate", "recommended_value"
].iloc[0]

THAI_RDI_FIBER = standard_df.loc[
    standard_df["metric"] == "Dietary Fiber", "recommended_value"
].iloc[0]


DGA_ADDED_SUGAR_LIMIT_PER_MEAL = dga_standard_df.loc[
    dga_standard_df["metric"] == "added_sugar_limit_per_meal", "max_value"
].iloc[0]
```

```python
df_model['sodium_pct_daily'] = (df_model['sodium'] / THAI_RDI_SODIUM) * 100
df_model['cholesterol_pct_daily'] = (df_model['cholesterol'] / THAI_RDI_CHOLESTEROL) * 100
df_model['sat_fat_pct_daily'] = (df_model['sat_fat'] / THAI_RDI_SAT_FAT) * 100
df_model['fat_pct_daily'] = (df_model['fat'] / THAI_RDI_FAT) * 100
df_model['carbs_pct_daily'] = (df_model['carbs'] / THAI_RDI_CARBS) * 100
df_model['fiber_pct_daily'] = (df_model['fiber'] / THAI_RDI_FIBER) * 100
df_model['sugar_pct_meal_limit'] = (df_model['sugar'] / DGA_ADDED_SUGAR_LIMIT_PER_MEAL) * 100

df_model.head()
```

Feature ใหม่เหล่านี้ช่วยให้สามารถตีความข้อมูลอาหารในเชิงมาตรฐานสุขภาพได้ง่ายขึ้น
เช่น sodium_pct_daily แสดงว่าอาหารหนึ่งรายการคิดเป็นกี่เปอร์เซ็นต์ของ sodium ต่อวันที่แนะนำ

```python
def create_risk_label(row):
    score = 0

    if row['sodium'] >= THAI_RDI_SODIUM:
        score += 1
    if row['sat_fat'] >= THAI_RDI_SAT_FAT:
        score += 1
    if row['cholesterol'] >= THAI_RDI_CHOLESTEROL:
        score += 1
    if row['sugar'] >= DGA_ADDED_SUGAR_LIMIT_PER_MEAL:
        score += 1

    if score >= 3:
        return 'High'
    elif score >= 1:
        return 'Medium'
    return 'Low'
```

```python
df_model['risk_level'] = df_model.apply(create_risk_label, axis=1)
df_model['risk_level'].value_counts()
```

```python
import seaborn as sns
import matplotlib.pyplot as plt

sns.countplot(x=df_model["risk_level"], order=["Low", "Medium", "High"])
plt.title("Risk Level Distribution")
plt.show()
```

```python
label_map = {'Low': 0, 'Medium': 1, 'High': 2}
df_model['risk_label'] = df_model['risk_level'].map(label_map)

df_model.head()
```

ในขั้นนี้ได้สร้าง target variable คือ risk_level โดยอาศัยกฎจากเกณฑ์โภชนาการ
เพื่อใช้เป็น baseline target สำหรับการทดลองสร้าง classification model

```python
def evaluate_food_item(row):
    messages = []

    if row["sodium"] >= THAI_RDI_SODIUM:
        messages.append("โซเดียมถึงหรือเกินเกณฑ์อ้างอิงต่อวัน")
    if row["cholesterol"] >= THAI_RDI_CHOLESTEROL:
        messages.append("โคเลสเตอรอลถึงหรือเกินเกณฑ์อ้างอิงต่อวัน")
    if row["sugar"] >= DGA_ADDED_SUGAR_LIMIT_PER_MEAL:
        messages.append("น้ำตาลถึงหรือเกินเกณฑ์ต่อมื้อ")
    if row["sat_fat"] >= THAI_RDI_SAT_FAT:
        messages.append("ไขมันอิ่มตัวถึงหรือเกินเกณฑ์อ้างอิงต่อวัน")

    if not messages:
        messages.append("ยังไม่พบตัวชี้วัดที่ถึงหรือเกินเกณฑ์อ้างอิง")

    return " | ".join(messages)
```

```python
df_model['chatbot_summary'] = df_model.apply(evaluate_food_item, axis=1)

if 'food_name' in df_model.columns:
    df_model[['food_name', 'risk_level', 'chatbot_summary']].head(10)
else:
    df_model[['risk_level', 'chatbot_summary']].head(10)
```

## Fun Testing: Q&A Demo

ส่วนนี้เป็นการทดลองถามตอบระบบเบื้องต้น
เพื่อดูว่าระบบสามารถตอบคำถามจาก knowledge, food risk และ recommendation ได้หรือไม่

```python
def ask_food_model(food_name, df):
    if "food_name" not in df.columns:
        return "ไม่พบคอลัมน์ food_name ใน df_model"

    result = df[df["food_name"].astype(str).str.lower() == food_name.lower()]

    if result.empty:
        return f"ไม่พบข้อมูลอาหารชื่อ {food_name}"

    row = result.iloc[0]

    lines = []
    lines.append(f"Food: {row['food_name']}")
    lines.append(f"Risk Level: {row['risk_level']}")
    lines.append(f"Summary: {row['chatbot_summary']}")
    lines.append(f"Calories: {row['calories']}")
    lines.append(f"Sugar: {row['sugar']}")
    lines.append(f"Cholesterol: {row['cholesterol']}")
    lines.append(f"Sodium: {row['sodium']}")
    lines.append(f"Saturated Fat: {row['sat_fat']}")

    return "\n".join(lines)
```

```python
test_foods = [
    "cream cheese",
    "neufchatel cheese",
    "ricotta cheese"
]

for food in test_foods:
    print("=" * 80)
    print(ask_food_model(food, df_model))
    print()
```

```python
def ask_risk_level(food_name, df):
    result = df[df["food_name"].astype(str).str.lower() == food_name.lower()]

    if result.empty:
        return f"ไม่พบข้อมูลอาหารชื่อ {food_name}"

    row = result.iloc[0]
    return f"{row['food_name']} มี risk level = {row['risk_level']} | {row['chatbot_summary']}"
```

```python
print(ask_risk_level("cream cheese", df_model))
print(ask_risk_level("ricotta cheese", df_model))
```

```python
def simple_food_qa(user_query, df):
    q = user_query.lower()

    if "ชื่อ" in q:
        return "กรุณาระบุชื่ออาหาร"

    if "เสี่ยงไหม" in q or "risk" in q or "สรุป" in q:
        if "food_name" not in df.columns:
            return "ไม่พบคอลัมน์ food_name"

        for food in df["food_name"].dropna().astype(str).unique():
            if food.lower() in q:
                row = df[df["food_name"].astype(str).str.lower() == food.lower()].iloc[0]
                return (
                    f"{row['food_name']}\n"
                    f"Risk Level: {row['risk_level']}\n"
                    f"Summary: {row['chatbot_summary']}\n"
                    f"Sugar: {row['sugar']}\n"
                    f"Cholesterol: {row['cholesterol']}\n"
                    f"Sodium: {row['sodium']}\n"
                    f"Saturated Fat: {row['sat_fat']}"
                )

        return "ไม่พบชื่ออาหารในคำถาม"

    return "ตอนนี้ระบบ demo นี้รองรับคำถามเกี่ยวกับความเสี่ยงและสรุปอาหาร"
```

```python
label_to_text = {
    0: "Low",
    1: "Medium",
    2: "High"
}
```

```python
def model_answer_food(food_name, df, model, feature_cols):
    if "food_name" not in df.columns:
        return "ไม่พบคอลัมน์ food_name ในข้อมูล"

    row = df[df["food_name"].astype(str).str.lower() == food_name.lower()]

    if row.empty:
        return f"ไม่พบข้อมูลอาหารชื่อ {food_name}"

    X_input = row[feature_cols]
    pred = model.predict(X_input)[0]
    risk_text = label_to_text.get(pred, str(pred))

    return f"โมเดลประเมินว่า {food_name} อยู่ในระดับความเสี่ยง {risk_text}"
```

```python
print(model_answer_food("cream", df_model_filtered, rf_pipeline, feature_cols))
print(model_answer_food("ricotta cheese", df_model_filtered, dt_pipeline, feature_cols))
print(model_answer_food("neufchatel cheese", df_model_filtered, lr_pipeline, feature_cols))
```

```python
def model_answer_food_detail(food_name, df, model, feature_cols):
    if "food_name" not in df.columns:
        return "ไม่พบคอลัมน์ food_name ในข้อมูล"

    row = df[df["food_name"].astype(str).str.lower() == food_name.lower()]

    if row.empty:
        return f"ไม่พบข้อมูลอาหารชื่อ {food_name}"

    row_data = row.iloc[0]
    X_input = row[feature_cols]
    pred = model.predict(X_input)[0]
    risk_text = label_to_text.get(pred, str(pred))

    response = [
        f"Food: {row_data['food_name']}",
        f"Predicted Risk Level: {risk_text}",
        f"Calories: {row_data['calories']}",
        f"Sugar: {row_data['sugar']}",
        f"Cholesterol: {row_data['cholesterol']}",
        f"Sodium: {row_data['sodium']}",
        f"Saturated Fat: {row_data['sat_fat']}",
    ]

    return "\n".join(response)
```
